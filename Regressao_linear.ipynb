Objetivo do Projeto
Desenvolver um modelo preditivo para estimar antecipadamente o desempenho (notas) de cursos universitários no Enade, permitindo que a instituição passe de uma postura reativa para uma estratégia preventiva.

O Problema de Negócio
Empresa/Cliente: A EduAnalytics foi contratada por uma grande rede de universidades.

Cenário Atual: A rede aplica ações de apoio (monitorias, plantões de dúvidas, trilhas) apenas depois que um curso obtém uma nota baixa no Enade.

Necessidade: Mudar essa abordagem reativa para prever as notas antes da realização do exame e agir antecipadamente.

A Solução Proposta
Modelagem Preditiva: Criar e otimizar modelos de Regressão Linear (simples e múltipla).

Análise de Fatores: Identificar quais variáveis socioeducacionais, acadêmicas e comportamentais mais impactam a nota final.

Ação Prática: Oferecer dados para que a rede elabore intervenções pedagógicas personalizadas antes da prova, evitando notas baixas.
